# Fair Compensation Project (Emily's Part)

We created a separate scaled feature set for clustering where columns like fairness index and fairness label were removed. Since, unlike regression (which is a supervised learning task), clustering is unsupervised, we did not split the dataset used for clustering. 

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import joblib
import os

df = pd.read_csv('data/cleaned/modelling_features.csv')
print(f"Starting shape: {df.shape}")
```

```python
print(df.columns.tolist())

drop_cols = [
    'fairness_index',
    'fairness_label',
    'job_id'
]

X = df.drop(columns = drop_cols)
```

```python
cols_to_scale = [
    'skills_count', 'total_employed', 'bls_annual_mean', 'bls_annual_median', 'bls_p10', 'bls_p90', 'pce_2023_in_millions', 'annual_mean', 'annual_low', 'annual_high'
]

cols_to_impute = ['total_employed', 'bls_p90']

# fit medians on data
medians = X[cols_to_impute].median()

print("Medians used for imputation:")
print(medians)

# apply
X[cols_to_impute] = X[cols_to_impute].fillna(medians)
```

## Standardize Numerical Values

Next, we scaled the numeric columns using StandardScaler. This standardized the values in the numeric columns so that no one column would have an outsized impact on the output of the model.

```python
scaler = StandardScaler()

# fit ONLY on training data
scaler.fit(X[cols_to_scale])

# transform both train and test
X[cols_to_scale] = scaler.transform(X[cols_to_scale])

```

```python
os.makedirs('data/clustering', exist_ok=True)
X.to_csv('data/cleaned/clustering/features_scaled.csv', index=False)
joblib.dump(scaler, 'models/scaler.pkl')
```
